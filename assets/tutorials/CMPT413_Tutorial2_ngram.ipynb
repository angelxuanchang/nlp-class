Code for constructing unigram and bigram language models

Below is our corpus:

```python
corpus = ["i love you", "i love him", "he love you", "he love me", "you love me"]
```

Create a Unigram model 

```python
unigrams = {}
total_unigrams = 0
for sentence in corpus:
    words = sentence.split(" ")
    total_unigrams += len(words)
    for word in words:
        if word not in unigrams:
            unigrams[word] = 1
        else:
            unigrams[word] += 1

for k, v in unigrams.items():
    unigrams[k] /= total_unigrams

print(unigrams)
```

Create a bigram model

```python
bigrams = {}
total_bigrams = 0
for sentence in corpus:
    words = sentence.split(" ")
    # We want to consider the EOP
    total_bigrams += len(words) + 1
    for i in range(len(words)+1):
        if i == 0:
            bigram = (None, words[i])
        elif i == len(words):
            bigram = (words[i-1], None)
        else:
            bigram = (words[i-1], words[i])
        if bigram not in bigrams:
            bigrams[bigram] = 1
        else:
            bigrams[bigram] += 1

for k, v in bigrams.items():
    bigrams[k] /= total_bigrams

print(bigrams)
```

Use the unigram model to calculate the probability of the sentence

```python
sent = "you love me"
prob = 1
words = sent.split(" ")
for word in words:
    prob *= unigrams[word]

print(prob)
```

Use the bigram model to calcualte the probability of the sentence

```python
sent = "you love him"
prob = 1
words = sent.split(" ")
for i in range(len(words) + 1):
    if i == 0:
        prob *= bigrams[(None, words[i])]
    elif i == len(words):
        prob *= bigrams[(words[i-1], None)]
    else:
        prob *= bigrams[(words[i-1], words[i])]

print(prob)
```
